```python
import serial

#arduinoData = serial.Serial('/dev/ttyACM1', 9600)
arduinoData = serial.Serial('COM3', 115200, timeout=0)

while True:
    myData = input('Kirimkan perintah (1/0): ')  
    if myData == "a":
        arduinoData.write(b'1')
    elif myData == "b":  
        arduinoData.write(b'0')
```
